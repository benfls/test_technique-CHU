```python
import pandas as pd
import numpy as np
```

```python
actors_df = pd.read_csv("../data/imdb_ijs_actors.tsv.gz", sep="\t")
roles_df = pd.read_csv("../data/imdb_ijs_roles.tsv.gz", sep="\t")
```

### Suppression des ID pairs

```python
actors_pair_df = actors_df[actors_df.id % 2 == 0]
```

```python
actors_pair_df.head()
```

### Jointure des 2 tables

```python
df_merge = pd.merge(actors_pair_df, roles_df, left_on='id', right_on='actor_id', how='inner')
```

```python
df_merge.head()
```

```python
df_merge.info()
```

### Count top 100 actor

```python
df_count = (
        df_merge
        .groupby(['id', 'first_name', 'last_name'], as_index=False)
        .size()
        .rename(columns={'size': 'role_count'})
        .sort_values(by='role_count', ascending=False)
        .head(100)
    )
```
